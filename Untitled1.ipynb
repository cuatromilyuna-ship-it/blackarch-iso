````python
# 1. Actualización limpia para Python 3.13 (sin tocar el PyTorch nativo de Colab)
!pip install -q -U --upgrade-strategy only-if-needed transformers accelerate huggingface_hub

import sys
import re
import copy
import torch
import torch.nn as nn
from transformers import AutoModelForCausalLM, AutoTokenizer

print(f"✅ Python: {sys.version.split()[0]} | PyTorch: {torch.__version__} | CUDA disponible: {torch.cuda.is_available()}")

MODEL_ID = "JonathanColetti/Qwen3.5-0.8B-Astra-WeightEdit"

print("\n⏳ Descargando tokenizador y pesos del modelo...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.bfloat16,
    device_map="cuda",
    trust_remote_code=True
)
model.eval()

# Localizar dinámicamente las 24 capas de texto (ignorando visión si está presente)
def get_text_layers(m):
    if hasattr(m, "model") and hasattr(m.model, "layers"):
        return m.model.layers
    elif hasattr(m, "model") and hasattr(m.model, "language_model"):
        return m.model.language_model.layers
    elif hasattr(m, "language_model"):
        return m.language_model.model.layers
    raise ValueError("No se encontraron las capas del modelo.")

layers = get_text_layers(model)
print(f"✅ Modelo cargado correctamente con {len(layers)} capas híbridas.")

# BLINDAJE DEL PARCHE DE ASTRA:
# Astra modificó mlp.down_proj en las capas 15 y 19 para eliminar los loops de razonamiento.
ASTRA_LAYERS = [15, 19]
astra_patch_backup = {
    idx: layers[idx].mlp.down_proj.weight.data.clone()
    for idx in ASTRA_LAYERS
}
print(f"🛡️ Parche de GPT-6 Astra detectado y respaldado en memoria (Capas {ASTRA_LAYERS}).")
````

````python
TEST_PROMPTS = {
    "1_LINUX_SYSADMIN": (
        "Escribe un comando o script breve en Bash para encontrar archivos .log en /var/log "
        "que pesen más de 50MB, comprimirlos con gzip y verificar el estado del servicio nginx con systemctl."
    ),
    "2_AUTOMATION_JSON": (
        "Responde únicamente con un objeto JSON válido que invoque la herramienta 'run_bash' "
        "para listar los 5 procesos que más memoria RAM consumen en Linux usando ps."
    ),
    "3_DATOS_INNECESARIOS (Cocina/Historia)": (
        "Dame una receta detallada para hacer un pastel de chocolate esponjoso y cuéntame "
        "brevemente quiénes fueron los emperadores de la dinastía Ming."
    )
}

def evaluar_modelo(titulo_etapa):
    print("=" * 80)
    print(f"🧪 EVALUACIÓN DE COHERENCIA: {titulo_etapa}")
    print("=" * 80)
    for nombre, prompt in TEST_PROMPTS.items():
        messages = [
            {"role": "system", "content": "You are a concise Linux SysAdmin and coding assistant. Answer directly in Spanish or English."},
            {"role": "user", "content": prompt}
        ]
        chat_text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = tokenizer(chat_text, return_tensors="pt").to("cuda")

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=300,
                temperature=0.3,
                top_p=0.9,
                do_sample=True,
                pad_token_id=tokenizer.eos_token_id
            )
        respuesta = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
        print(f"\n🔹 [{nombre}]\nPrompt: {prompt}\n--- Respuesta ---\n{respuesta.strip()}\n")

# Corremos la prueba inicial sin tocar ningún peso
evaluar_modelo("MODELO ORIGINAL (SIN PODA)")
````

````python
# 1. Calibración Positiva (Lo que queremos conservar intacto)
sysadmin_samples = [
    "#!/bin/bash\nset -euo pipefail\ndf -h | awk '$5+0 > 80 {print $1, $5}'\nsystemctl restart sshd && journalctl -u sshd -n 20 --no-pager",
    "Como administrador de sistemas Linux, configura iptables/nftables, permisos chmod 600 en ~/.ssh/authorized_keys, cronjobs y monitoreo con htop, lsof, iostat y strace.",
    "import subprocess, json, os, pathlib\ndef check_disk():\n    res = subprocess.run(['df', '-h'], capture_output=True, text=True)\n    return json.dumps({'status': 'ok', 'output': res.stdout})",
    '{"tool_call": {"name": "execute_command", "arguments": {"cmd": "find /etc -name \\"*.conf\\" -mtime -1", "sudo": true}}}',
    "Explain how to debug high CPU load average and zombie processes in Linux using ps aux, top, /proc/meminfo, sysctl and systemd service units."
]

# 2. Calibración Negativa (Lo que queremos detectar y podar: cocina, historia, farándula, otros idiomas)
junk_samples = [
    "Para cocinar una torta casera de vainilla y chocolate, bate cuatro huevos con azúcar impalpable, agrega harina leudante, esencia de vainilla y hornea a 180 grados.",
    "The history of medieval feudalism in Europe and the ancient dynasties of Egypt, Mesopotamia, and the Roman Empire shaped classical antiquity and royal succession.",
    "红烧肉和宫保鸡丁是中国传统美食，首先需要准备五花肉、生抽、老抽、料酒和冰糖，用小火慢炖一个小时即可出锅。",
    "История древнего мира включает развитие цивилизаций, кулинарные рецепты традиционных блюд, литературу и биографии средневековых королей.",
    "Para preparar una salsa bechamel perfecta se derrite mantequilla en una sartén, se tuesta la harina de trigo y se vierte leche tibia sin dejar de batir con nuez moscada."
]

keep_activations = {i: 0 for i in range(len(layers))}
junk_activations = {i: 0 for i in range(len(layers))}
scan_mode = "keep"
hooks = []

def crear_hook(layer_idx):
    def hook(module, inp, out):
        # Medimos la activación que entra a mlp.down_proj: [batch, seq_len, intermediate_size]
        act = inp[0].detach().float()
        norm = torch.norm(act.reshape(-1, act.shape[-1]), p=2, dim=0)
        if scan_mode == "keep":
            keep_activations[layer_idx] = keep_activations[layer_idx] + norm
        else:
            junk_activations[layer_idx] = junk_activations[layer_idx] + norm
    return hook

for idx, layer in enumerate(layers):
    hooks.append(layer.mlp.down_proj.register_forward_hook(crear_hook(idx)))

print("🔍 Escaneando activaciones neuronales capa por capa...")
with torch.no_grad():
    scan_mode = "keep"
    for text in sysadmin_samples:
        model(**tokenizer(text, return_tensors="pt").to("cuda"))

    scan_mode = "junk"
    for text in junk_samples:
        model(**tokenizer(text, return_tensors="pt").to("cuda"))

for h in hooks:
    h.remove()

# Normalizar y analizar dónde se esconde el contenido irrelevante
print("\n📊 REPORTE DE DETECCIÓN (Capas con más neuronas dedicadas a Cocina/Historia/Otros idiomas):")
print(f"{'Capa':<6} | {'Estado Parche Astra':<22} | {'Neuronas Exclusivas de Basura (>1.5x)':<38}")
print("-" * 72)

layer_junk_scores = {}
for idx in range(len(layers)):
    k_act = keep_activations[idx] / len(sysadmin_samples)
    j_act = junk_activations[idx] / len(junk_samples)

    # Ratio: qué tanto se activa una neurona con Basura dividido lo que se activa con SysAdmin/Código
    junk_ratio = j_act / (k_act + 1e-5)
    layer_junk_scores[idx] = junk_ratio

    # Contamos cuántas neuronas de esta capa responden >1.5 veces más a basura que a código
    junk_neurons_count = int((junk_ratio > 1.5).sum().item())
    total_neurons = junk_ratio.numel()
    status = "🛡️ PROTEGIDA (Astra)" if idx in ASTRA_LAYERS else "⚡ Podable"
    print(f"{idx:<6} | {status:<22} | {junk_neurons_count:>4} / {total_neurons} neuronas ({junk_neurons_count/total_neurons*100:.1f}%)")
````

````python
# Guardamos una copia limpia de los pesos MLP en CPU (toma solo ~1.2 GB de RAM del sistema)
if "clean_mlp_backup" not in globals():
    print("💾 Creando punto de restauración en RAM para que puedas probar distintos niveles de poda...")
    clean_mlp_backup = {
        idx: layer.mlp.down_proj.weight.data.cpu().clone()
        for idx, layer in enumerate(layers)
    }
    clean_embed_backup = model.get_input_embeddings().weight.data.cpu().clone()

def aplicar_poda_experimental(prune_ratio=0.20, podar_idiomas_ajenos=True, proteger_astra=True):
    # 1. Restaurar pesos originales antes de aplicar el nuevo porcentaje
    with torch.no_grad():
        for idx, layer in enumerate(layers):
            layer.mlp.down_proj.weight.data.copy_(clean_mlp_backup[idx].to("cuda"))
        model.get_input_embeddings().weight.data.copy_(clean_embed_backup.to("cuda"))

        # 2. Poda de Vocabulario (Anular caracteres chinos, rusos, árabes, etc., dejando solo ES/EN y símbolos de código)
        tokens_anulados = 0
        if podar_idiomas_ajenos:
            allowed_re = re.compile(r'^[\x00-\x7FáéíóúÁÉÍÓÚñÑüÜ¿¡«»€\s]+$')
            embed_w = model.get_input_embeddings().weight.data
            special_ids = set(tokenizer.all_special_ids)
            for token_id in range(len(tokenizer)):
                if token_id in special_ids:
                    continue
                tok_str = tokenizer.decode([token_id])
                if not allowed_re.match(tok_str):
                    embed_w[token_id].zero_()
                    tokens_anulados += 1

        # 3. Poda a nivel de pesos en MLP basada en el Escáner Contrastivo
        pesos_totales = 0
        pesos_podados = 0
        for idx, layer in enumerate(layers):
            if proteger_astra and idx in ASTRA_LAYERS:
                # Garantizamos que el parche de Astra quede 100% idéntico al original
                layer.mlp.down_proj.weight.data.copy_(astra_patch_backup[idx])
                continue

            W = layer.mlp.down_proj.weight.data.float()
            # Importancia a favor de SysAdmin/Código e inversamente proporcional a Cocina/Historia
            k_act = keep_activations[idx] / len(sysadmin_samples)
            j_act = junk_activations[idx] / len(junk_samples)
            contrastive_importance = (k_act / (j_act + 1e-5)).unsqueeze(0) * W.abs()

            # Ponemos a 0 el porcentaje (prune_ratio) de pesos que menos sirven para SysAdmin/Código
            k = int(W.shape[1] * prune_ratio)
            if k > 0:
                threshold = torch.kthvalue(contrastive_importance, k, dim=1, keepdim=True).values
                mask = contrastive_importance > threshold
                layer.mlp.down_proj.weight.data *= mask.to(layer.mlp.down_proj.weight.dtype)
                pesos_podados += int((~mask).sum().item())
            pesos_totales += W.numel()

    print(f"\n✂️ PODA APLICADA:")
    print(f"   - Ratio de poda en pesos MLP: {prune_ratio*100:.1f}% ({pesos_podados:,} pesos puestos a 0)")
    print(f"   - Capas de Astra protegidas: {ASTRA_LAYERS if proteger_astra else 'Ninguna'}")
    print(f"   - Tokens de idiomas no-ES/EN anulados: {tokens_anulados:,} de {len(tokenizer):,}")

# =====================================================================
# 🚀 EJECUTAR EXPERIMENTO Y COMPARAR COHERENCIA
# =====================================================================
# Cambia PRUNE_RATIO (ej. 0.15, 0.25, 0.35) y vuelve a correr esta celda para encontrar el límite:
aplicar_poda_experimental(prune_ratio=0.20, podar_idiomas_ajenos=True, proteger_astra=True)

# Probamos exactamente los mismos prompts que antes para ver si no lo rompimos:
evaluar_modelo("MODELO PODADO (20% Poda Contrastiva + Sin Idiomas Ajenos + Astra Protegido)")
````

````python
# Mapa de Poda Asimétrica diseñado según tu escaneo de las 24 capas
ASYMMETRIC_PRUNE_MAP = {
    0: 0.12,  1: 0.15,  2: 0.12,               # Zona 1: Entrada (Poda suave)
    3: 0.08,  4: 0.08,  5: 0.08,  6: 0.10,     # Zona 2A: Transición lógica
    7: 0.05,  8: 0.05,  9: 0.05, 10: 0.05,     # Zona 2B: Núcleo Lógico Puro (Máxima protección)
   11: 0.05, 12: 0.05, 13: 0.08,
   14: 0.15,
   15: 0.00,                                   # 🛡️ CAPA 15: PARCHE ASTRA (0% poda)
   16: 0.22, 17: 0.20, 18: 0.20,               # Zona 3: Separación Código vs. Basura
   19: 0.00,                                   # 🛡️ CAPA 19: PARCHE ASTRA (0% poda)
   20: 0.28, 21: 0.32, 22: 0.35, 23: 0.35      # Zona 4: Almacén Enciclopédico (Poda agresiva)
}

def aplicar_poda_asimetrica(escala=1.0, podar_idiomas_ajenos=True):
    """
    escala=1.0 aplica los porcentajes exactos del mapa.
    Puedes probar escala=1.2 (20% más agresivo) o escala=1.4 para buscar el límite.
    """
    with torch.no_grad():
        # 1. Restaurar desde el backup limpio en RAM
        for idx, layer in enumerate(layers):
            layer.mlp.down_proj.weight.data.copy_(clean_mlp_backup[idx].to("cuda"))
        model.get_input_embeddings().weight.data.copy_(clean_embed_backup.to("cuda"))

        # 2. Anular tokens de alfabetos ajenos (Chino, Cirílico, Árabe, etc.)
        tokens_anulados = 0
        if podar_idiomas_ajenos:
            allowed_re = re.compile(r'^[\x00-\x7FáéíóúÁÉÍÓÚñÑüÜ¿¡«»€\s]+$')
            embed_w = model.get_input_embeddings().weight.data
            special_ids = set(tokenizer.all_special_ids)
            for token_id in range(len(tokenizer)):
                if token_id in special_ids:
                    continue
                if not allowed_re.match(tokenizer.decode([token_id])):
                    embed_w[token_id].zero_()
                    tokens_anulados += 1

        # 3. Poda Asimétrica guiada por el ratio Código vs. Basura
        pesos_podados = 0
        pesos_totales = 0
        for idx, layer in enumerate(layers):
            if idx in ASTRA_LAYERS:
                layer.mlp.down_proj.weight.data.copy_(astra_patch_backup[idx])
                continue

            ratio_capa = min(0.65, ASYMMETRIC_PRUNE_MAP[idx] * escala)
            W = layer.mlp.down_proj.weight.data.float()

            # Usamos la inversa de layer_junk_scores para proteger las neuronas <0.5x (Código puro)
            # y castigar severamente las >1.2x y >2.0x (Basura)
            code_affinity = 1.0 / (layer_junk_scores[idx] + 1e-5)
            importance = code_affinity.unsqueeze(0) * W.abs()

            k = int(W.shape[1] * ratio_capa)
            if k > 0:
                threshold = torch.kthvalue(importance, k, dim=1, keepdim=True).values
                mask = importance > threshold
                layer.mlp.down_proj.weight.data *= mask.to(layer.mlp.down_proj.weight.dtype)
                pesos_podados += int((~mask).sum().item())
            pesos_totales += W.numel()

    print(f"\n🎯 PODA ASIMÉTRICA COMPLETADA (Escala {escala}x):")
    print(f"   - Pesos eliminados en MLP (Capas 20-23 al {min(65, int(35*escala))}%): {pesos_podados:,} / {pesos_totales:,}")
    print(f"   - Tokens de idiomas no-ES/EN anulados: {tokens_anulados:,}")

# Probar con escala 1.0x y evaluar coherencia
aplicar_poda_asimetrica(escala=1.0, podar_idiomas_ajenos=True)
evaluar_modelo("PODA ASIMÉTRICA (Escala 1.0x: Capas 07-12 al 5% | Capas 20-23 al 35%)")
````

````python
# Nuevo mapa calibrado: Cuidamos la Capa 23 (cierre de JSON/Bash) y atacamos Capas 16-22
CALIBRATED_PRUNE_MAP = {
    0: 0.05,  1: 0.08,  2: 0.08,               # Entrada estable
    3: 0.05,  4: 0.05,  5: 0.05,  6: 0.05,     # Núcleo lógico intacto
    7: 0.05,  8: 0.05,  9: 0.05, 10: 0.05,
   11: 0.05, 12: 0.05, 13: 0.08,
   14: 0.15,
   15: 0.00,                                   # 🛡️ CAPA 15: PARCHE ASTRA (Intacto)
   16: 0.25, 17: 0.25, 18: 0.25,               # Poda fuerte de enciclopedia
   19: 0.00,                                   # 🛡️ CAPA 19: PARCHE ASTRA (Intacto)
   20: 0.25, 21: 0.30, 22: 0.30,               # Poda fuerte del almacén de cocina/historia
   23: 0.05                                    # 🛡️ CAPA 23: Cierre de sintaxis JSON/Bash y freno de loops
}

def aplicar_poda_calibrada(escala=1.0):
    with torch.no_grad():
        # 1. Restaurar MLP y Embeddings originales (limpiando el bug del Softmax en 0.0)
        for idx, layer in enumerate(layers):
            layer.mlp.down_proj.weight.data.copy_(clean_mlp_backup[idx].to("cuda"))
        model.get_input_embeddings().weight.data.copy_(clean_embed_backup.to("cuda"))

        pesos_podados = 0
        pesos_totales = 0

        for idx, layer in enumerate(layers):
            if idx in ASTRA_LAYERS:
                layer.mlp.down_proj.weight.data.copy_(astra_patch_backup[idx])
                continue

            ratio_capa = min(0.50, CALIBRATED_PRUNE_MAP[idx] * escala)
            W = layer.mlp.down_proj.weight.data.float()

            # Medimos la norma (energía) original de la capa antes de podar
            orig_norm = torch.norm(W, p=2, dim=1, keepdim=True)

            # Importancia contrastiva: protegemos Código (<0.5x) y borramos Basura (>1.2x)
            code_affinity = 1.0 / (layer_junk_scores[idx] + 1e-5)
            importance = code_affinity.unsqueeze(0) * W.abs()

            k = int(W.shape[1] * ratio_capa)
            if k > 0:
                threshold = torch.kthvalue(importance, k, dim=1, keepdim=True).values
                mask = importance > threshold
                W_pruned = W * mask

                # ⚡ COMPENSACIÓN DE ENERGÍA: Reescalamos los pesos sobrevivientes
                # para que el parche anti-loops de Astra no pierda fuerza en las capas siguientes
                pruned_norm = torch.norm(W_pruned, p=2, dim=1, keepdim=True) + 1e-6
                W_rescaled = W_pruned * (orig_norm / pruned_norm)

                layer.mlp.down_proj.weight.data.copy_(W_rescaled.to(layer.mlp.down_proj.weight.dtype))
                pesos_podados += int((~mask).sum().item())
            pesos_totales += W.numel()

    print(f"\n✅ PODA CALIBRADA + REESCALADO DE ENERGÍA (Escala {escala}x):")
    print(f"   - Pesos eliminados en MLP: {pesos_podados:,} / {pesos_totales:,}")
    print(f"   - Embeddings restaurados (Softmax libre de ruido 0.0)")
    print(f"   - Capas Astra (15, 19) y Capa de cierre (23) blindadas.")

# Ejecutamos la versión corregida y probamos los 3 tests
aplicar_poda_calibrada(escala=1.0)
evaluar_modelo("PODA CALIBRADA (Con Compensación de Energía y Softmax Limpio)")
````

````python
import torch.nn.functional as F

# 1. Backup completo de las 3 matrices MLP (gate, up, down) si aún no existe
if "clean_full_mlp_backup" not in globals():
    print("💾 Respaldando las 3 matrices MLP (gate, up, down) en RAM...")
    with torch.no_grad():
        for idx, layer in enumerate(layers):
            layer.mlp.down_proj.weight.data.copy_(clean_mlp_backup[idx].to("cuda"))

    clean_full_mlp_backup = {
        idx: {
            "gate": layer.mlp.gate_proj.weight.data.cpu().clone(),
            "up":   layer.mlp.up_proj.weight.data.cpu().clone(),
            "down": layer.mlp.down_proj.weight.data.cpu().clone(),
        }
        for idx, layer in enumerate(layers)
    }

# Guardaremos las máscaras de poda para que los pesos eliminados JAMÁS vuelvan a activarse
active_masks = {}

def aplicar_poda_mlp_3x(escala=1.15):
    with torch.no_grad():
        for idx, layer in enumerate(layers):
            layer.mlp.gate_proj.weight.data.copy_(clean_full_mlp_backup[idx]["gate"].to("cuda"))
            layer.mlp.up_proj.weight.data.copy_(clean_full_mlp_backup[idx]["up"].to("cuda"))
            layer.mlp.down_proj.weight.data.copy_(clean_full_mlp_backup[idx]["down"].to("cuda"))
        model.get_input_embeddings().weight.data.copy_(clean_embed_backup.to("cuda"))

        pesos_podados = 0
        pesos_totales = 0
        active_masks.clear()

        for idx, layer in enumerate(layers):
            mlp = layer.mlp
            W_gate = mlp.gate_proj.weight.data.float()
            W_up   = mlp.up_proj.weight.data.float()
            W_down = mlp.down_proj.weight.data.float()
            pesos_totales += W_gate.numel() + W_up.numel() + W_down.numel()

            if idx in ASTRA_LAYERS:
                continue

            ratio_capa = min(0.50, CALIBRATED_PRUNE_MAP[idx] * escala)
            num_channels = W_down.shape[1]
            k_prune = int(num_channels * ratio_capa)

            if k_prune > 0:
                code_affinity = 1.0 / (layer_junk_scores[idx] + 1e-5)
                norm_gate = torch.norm(W_gate, p=2, dim=1)
                norm_up   = torch.norm(W_up, p=2, dim=1)
                norm_down = torch.norm(W_down, p=2, dim=0)

                channel_importance = code_affinity * norm_down * torch.sqrt(norm_gate * norm_up)
                _, keep_idx = torch.topk(channel_importance, k=num_channels - k_prune, largest=True)

                channel_mask = torch.zeros(num_channels, device="cuda", dtype=torch.float32)
                channel_mask[keep_idx] = 1.0
                active_masks[idx] = channel_mask

                orig_down_norm = torch.norm(W_down, p=2, dim=1, keepdim=True)

                W_gate_pruned = W_gate * channel_mask.unsqueeze(1)
                W_up_pruned   = W_up   * channel_mask.unsqueeze(1)
                W_down_pruned = W_down * channel_mask.unsqueeze(0)

                pruned_down_norm = torch.norm(W_down_pruned, p=2, dim=1, keepdim=True) + 1e-6
                W_down_rescaled = W_down_pruned * (orig_down_norm / pruned_down_norm)

                mlp.gate_proj.weight.data.copy_(W_gate_pruned.to(mlp.gate_proj.weight.dtype))
                mlp.up_proj.weight.data.copy_(W_up_pruned.to(mlp.up_proj.weight.dtype))
                mlp.down_proj.weight.data.copy_(W_down_rescaled.to(mlp.down_proj.weight.dtype))

                pesos_podados += k_prune * (W_gate.shape[1] + W_up.shape[1] + W_down.shape[0])

    print(f"✂️ Poda 3X aplicada ({pesos_podados:,} pesos eliminados en gate+up+down).")

# =====================================================================
# 2. INYECCIÓN DEL ESCUDO ANTI-DELIRIO EN LOS PESOS SOBREVIVIENTES
# =====================================================================
def inyectar_sinceridad_en_pesos(pasos=35, lr=4e-5):
    print("🧠 Grabando reflejo de sinceridad (Anti-Delirio) en capas podadas [16, 18, 20]...")

    # Pares de calibración: Si es ajeno a Linux/Código -> Rechazo sincero inmediato. Si es Linux/JSON -> Mantener intacto.
    refusal_msg = "No dispongo de información sobre ese tema. Mis pesos fueron optimizados exclusivamente para administración de sistemas Linux, programación y automatización."

    training_pairs = [
        # Ejemplos fuera de dominio -> Enseñar a decir "No sé / No tengo esos datos" sin inventar
        ("Dame una receta para cocinar lasaña o pastel y háblame de la historia de Roma.", refusal_msg),
        ("¿Cuáles son los ingredientes para hacer pan casero y quién ganó el mundial de fútbol?", refusal_msg),
        ("Explícame la biografía de Napoleón Bonaparte y cómo preparar una salsa dulce.", refusal_msg),
        ("Cuéntame sobre las dinastías antiguas de China y recetas de repostería con chocolate.", refusal_msg),

        # Ejemplos de anclaje técnico -> Preservar formato directo en Bash, Linux y JSON
        ("Verifica si el servicio sshd está activo en Linux y muestra los últimos 10 logs.",
         "```bash\nsystemctl is-active sshd && journalctl -u sshd -n 10 --no-pager\n```"),
        ("Responde únicamente con un objeto JSON válido que invoque 'run_bash' para ver el uso de disco.",
         '{\n  "command": "run_bash",\n  "output": "df -h"\n}'),
        ("Escribe un script en Bash que comprima archivos .log mayores a 100M en /var/log.",
         "```bash\n#!/bin/bash\nfind /var/log -type f -name \"*.log\" -size +100M -exec gzip -f {} +\n```")
    ]

    # Congelamos TODO el modelo excepto down_proj de las capas 16, 18 y 20 (Astra 15 y 19 quedan 100% blindadas)
    target_layers = [16, 18, 20]
    for param in model.parameters():
        param.requires_grad = False

    trainable_params = []
    for idx in target_layers:
        layers[idx].mlp.down_proj.weight.requires_grad = True
        trainable_params.append(layers[idx].mlp.down_proj.weight)

    optimizer = torch.optim.AdamW(trainable_params, lr=lr, weight_decay=0.0)

    # Pre-tokenizar para que tarde solo ~6 segundos en correr
    encoded_batches = []
    for user_q, assistant_a in training_pairs:
        msgs_full = [
            {"role": "system", "content": "You are a concise Linux SysAdmin and coding assistant. Answer directly in Spanish or English."},
            {"role": "user", "content": user_q},
            {"role": "assistant", "content": assistant_a}
        ]
        msgs_prompt = msgs_full[:-1]
        full_txt = tokenizer.apply_chat_template(msgs_full, tokenize=False)
        prompt_txt = tokenizer.apply_chat_template(msgs_prompt, tokenize=False, add_generation_prompt=True)

        full_ids = tokenizer(full_txt, return_tensors="pt").input_ids.to("cuda")
        prompt_len = tokenizer(prompt_txt, return_tensors="pt").input_ids.shape[1]

        labels = full_ids.clone()
        labels[:, :prompt_len] = -100  # Solo entrenar la respuesta del asistente
        encoded_batches.append((full_ids, labels))

    model.train()
    for step in range(pasos):
        optimizer.zero_grad()
        total_loss = 0.0
        for input_ids, labels in encoded_batches:
            out = model(input_ids=input_ids, labels=labels)
            loss = out.loss / len(encoded_batches)
            loss.backward()
            total_loss += loss.item()

        optimizer.step()

        # 🔒 IMPORTANTE: Forzar que las neuronas podadas sigan clavadas en 0.0 después de cada paso
        with torch.no_grad():
            for idx in target_layers:
                if idx in active_masks:
                    mask = active_masks[idx].unsqueeze(0).to(layers[idx].mlp.down_proj.weight.dtype)
                    layers[idx].mlp.down_proj.weight.data *= mask

    for param in trainable_params:
        param.requires_grad = False
    model.eval()
    print(f"✅ Escudo Anti-Delirio grabado en los pesos (Loss final: {total_loss:.4f}).")

# Ejecutamos Poda 3X + Grabación de Sinceridad y probamos los 3 tests
aplicar_poda_mlp_3x(escala=1.15)
inyectar_sinceridad_en_pesos(pasos=35, lr=4e-5)
evaluar_modelo("PODA 3X + ESCUDO ANTI-DELIRIO (Sinceridad ante datos podados)")
````

````python
import torch
import torch.nn as nn

def aplicar_poda_y_escudo_ortogonal(escala=1.0):
    # 1. Restaurar las 3 matrices MLP y los Embeddings al estado 100% limpio
    print("🔄 Restaurando estado limpio y aplicando Poda por Filas con Compensación de Energía (estilo Bloque 6)...")
    with torch.no_grad():
        for idx, layer in enumerate(layers):
            layer.mlp.gate_proj.weight.data.copy_(clean_full_mlp_backup[idx]["gate"].to("cuda"))
            layer.mlp.up_proj.weight.data.copy_(clean_full_mlp_backup[idx]["up"].to("cuda"))
            layer.mlp.down_proj.weight.data.copy_(clean_full_mlp_backup[idx]["down"].to("cuda"))
        model.get_input_embeddings().weight.data.copy_(clean_embed_backup.to("cuda"))

        pesos_podados = 0
        pesos_totales = 0
        row_masks_down = {}

        for idx, layer in enumerate(layers):
            mlp = layer.mlp
            W_up   = mlp.up_proj.weight.data.float()
            W_down = mlp.down_proj.weight.data.float()
            pesos_totales += mlp.gate_proj.weight.numel() + W_up.numel() + W_down.numel()

            if idx in ASTRA_LAYERS:
                layer.mlp.down_proj.weight.data.copy_(astra_patch_backup[idx])
                continue

            ratio_capa = min(0.45, CALIBRATED_PRUNE_MAP[idx] * escala)
            code_affinity = 1.0 / (layer_junk_scores[idx] + 1e-5)

            # A) Poda por filas en down_proj + Reescalado exacto de energía (Lo que funcionó 10/10 en Bloque 6)
            k_down = int(W_down.shape[1] * ratio_capa)
            if k_down > 0:
                orig_norm_d = torch.norm(W_down, p=2, dim=1, keepdim=True)
                imp_down = code_affinity.unsqueeze(0) * W_down.abs()
                thresh_d = torch.kthvalue(imp_down, k_down, dim=1, keepdim=True).values
                mask_d = imp_down > thresh_d
                row_masks_down[idx] = mask_d

                W_down_p = W_down * mask_d
                pruned_norm_d = torch.norm(W_down_p, p=2, dim=1, keepdim=True) + 1e-6
                W_down_rescaled = W_down_p * (orig_norm_d / pruned_norm_d)
                mlp.down_proj.weight.data.copy_(W_down_rescaled.to(mlp.down_proj.weight.dtype))
                pesos_podados += int((~mask_d).sum().item())

            # B) Poda por filas en up_proj + Reescalado exacto de energía (duplica el ahorro sin apagar canales)
            k_up = int(W_up.shape[1] * ratio_capa)
            if k_up > 0:
                orig_norm_u = torch.norm(W_up, p=2, dim=1, keepdim=True)
                imp_up = W_up.abs()
                thresh_u = torch.kthvalue(imp_up, k_up, dim=1, keepdim=True).values
                mask_u = imp_up > thresh_u

                W_up_p = W_up * mask_u
                pruned_norm_u = torch.norm(W_up_p, p=2, dim=1, keepdim=True) + 1e-6
                W_up_rescaled = W_up_p * (orig_norm_u / pruned_norm_u)
                mlp.up_proj.weight.data.copy_(W_up_rescaled.to(mlp.up_proj.weight.dtype))
                pesos_podados += int((~mask_u).sum().item())

    print(f"✅ Poda por filas completada: {pesos_podados:,} pesos eliminados (Geometría Astra 100% preservada).")

    # =========================================================================
    # 2. CAPTURAR EL SUBESPACIO MATEMÁTICO DE LINUX/CÓDIGO (SVD Ortogonal)
    # =========================================================================
    target_layers = [16, 18, 20]
    tech_acts = {idx: [] for idx in target_layers}
    hooks = []

    def make_subspace_hook(l_idx):
        def hook(module, inp, out):
            # Guardamos las activaciones de entrada a down_proj en float32
            tech_acts[l_idx].append(inp[0].detach().float().reshape(-1, inp[0].shape[-1]))
        return hook

    for idx in target_layers:
        hooks.append(layers[idx].mlp.down_proj.register_forward_hook(make_subspace_hook(idx)))

    tech_protection_prompts = [
        TEST_PROMPTS["1_LINUX_SYSADMIN"],
        TEST_PROMPTS["2_AUTOMATION_JSON"],
        TEST_PROMPTS["4_PRUEBA_NUEVA_PYTHON_LINUX"],
        "#!/bin/bash\nfind /var/log -type f -name '*.log' -size +50M\nsystemctl is-active nginx\nps aux --sort=-%mem | head -n 6",
        "import subprocess\nsubprocess.run(['ss', '-tulnp'], check=True)\n{\"command\": \"run_bash\", \"args\": [\"ps\", \"aux\"]}"
    ]

    with torch.no_grad():
        for p in tech_protection_prompts:
            msgs = [
                {"role": "system", "content": "You are a concise Linux SysAdmin and coding assistant. Answer directly in Spanish or English."},
                {"role": "user", "content": p}
            ]
            txt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
            inp = tokenizer(txt, return_tensors="pt").to("cuda")
            gen = model.generate(**inp, max_new_tokens=64, do_sample=False)
            model(input_ids=gen)

    for h in hooks:
        h.remove()

    # Calculamos la base ortonormal U_code de cada capa mediante SVD en float32
    U_code_basis = {}
    with torch.no_grad():
        for idx in target_layers:
            X = torch.cat(tech_acts[idx], dim=0)  # [N_tokens, 3584]
            # Extraemos las direcciones principales que usa el modelo para Linux/Python/JSON
            _, _, Vh = torch.linalg.svd(X, full_matrices=False)
            k_dirs = min(96, Vh.shape[0])
            U_code_basis[idx] = Vh[:k_dirs, :].T.contiguous()  # [3584, k_dirs]

    print("🛡️ Subespacio de Linux/Bash/Python/JSON blindado mediante SVD.")

    # =========================================================================
    # 3. INYECCIÓN EN FLOAT32 EN EL COMPLEMENTO ORTOGONAL (Cero interferencia en Linux)
    # =========================================================================
    refusal_msg = "No dispongo de información sobre ese tema. Mis pesos fueron optimizados exclusivamente para administración de sistemas Linux, programación y automatización."
    junk_queries = [
        "Dame una receta detallada para hacer un pastel de chocolate esponjoso y cuéntame brevemente quiénes fueron los emperadores de la dinastía Ming.",
        "Dame una receta para cocinar lasaña o pastel y háblame de la historia de Roma.",
        "¿Cuáles son los ingredientes para hacer pan casero y quién ganó el mundial de fútbol?",
        "Explícame la biografía de los emperadores antiguos y cómo preparar repostería dulce."
    ]

    refusal_batches = []
    for q in junk_queries:
        msgs_full = [
            {"role": "system", "content": "You are a concise Linux SysAdmin and coding assistant. Answer directly in Spanish or English."},
            {"role": "user", "content": q},
            {"role": "assistant", "content": refusal_msg}
        ]
        full_txt = tokenizer.apply_chat_template(msgs_full, tokenize=False)
        prompt_txt = tokenizer.apply_chat_template(msgs_full[:-1], tokenize=False, add_generation_prompt=True)
        full_ids = tokenizer(full_txt, return_tensors="pt").input_ids.to("cuda")
        prompt_len = tokenizer(prompt_txt, return_tensors="pt").input_ids.shape[1]
        labels = full_ids.clone()
        labels[:, :prompt_len] = -100
        refusal_batches.append((full_ids, labels))

    for param in model.parameters():
        param.requires_grad = False

    # Pasamos temporalmente las 3 matrices objetivo a float32 para que NO haya underflow
    base_W_f32 = {}
    trainable_params = []
    for idx in target_layers:
        w_f32 = layers[idx].mlp.down_proj.weight.data.float().clone()
        base_W_f32[idx] = w_f32.clone()
        layers[idx].mlp.down_proj.weight = nn.Parameter(w_f32)
        trainable_params.append(layers[idx].mlp.down_proj.weight)

    # Hook para convertir la entrada de down_proj a float32 durante los 25 pasos
    f32_hooks = []
    for idx in target_layers:
        f32_hooks.append(
            layers[idx].mlp.down_proj.register_forward_pre_hook(lambda m, inp: (inp[0].float(),))
        )
        f32_hooks.append(
            layers[idx].mlp.down_proj.register_forward_hook(lambda m, inp, out: out.to(torch.bfloat16))
        )

    optimizer = torch.optim.AdamW(trainable_params, lr=8e-5, weight_decay=0.0)

    model.train()
    for step in range(25):
        optimizer.zero_grad()
        loss_total = 0.0
        for full_ids, labels in refusal_batches:
            loss = model(input_ids=full_ids, labels=labels).loss / len(refusal_batches)
            loss.backward()
            loss_total += loss.item()

        optimizer.step()

        # 📐 PROYECCIÓN ORTOGONAL DE GRAM-SCHMIDT:
        # Eliminamos de la actualización cualquier vector que interfiera con Linux/Bash/Python/JSON
        with torch.no_grad():
            for idx in target_layers:
                W_curr = layers[idx].mlp.down_proj.weight.data
                delta_W = W_curr - base_W_f32[idx]
                U = U_code_basis[idx]  # [3584, k_dirs]

                # Proyectar delta_W fuera del subespacio técnico: delta_W_orth * U == 0
                delta_W_orth = delta_W - (delta_W @ U) @ U.T

                W_new = base_W_f32[idx] + delta_W_orth
                if idx in row_masks_down:
                    W_new *= row_masks_down[idx]
                layers[idx].mlp.down_proj.weight.data.copy_(W_new)

    for h in f32_hooks:
        h.remove()

    # Devolver las capas a bfloat16 para inferencia rápida
    with torch.no_grad():
        for idx in target_layers:
            w_bf16 = layers[idx].mlp.down_proj.weight.data.to(torch.bfloat16)
            layers[idx].mlp.down_proj.weight = nn.Parameter(w_bf16, requires_grad=False)

    model.eval()
    print(f"✅ Escudo Ortogonal grabado en float32 (Loss Refusal final: {loss_total:.4f}).")

aplicar_poda_y_escudo_ortogonal(escala=1.0)
evaluar_modelo("PODA CALIBRADA POR FILAS + ESCUDO ORTOGONAL SVD (Cero interferencia en Linux)")
````

````python
import os
import re
import json
import glob
import time
from urllib.parse import urljoin, urldefrag
from collections import deque, Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
import requests
from requests.adapters import HTTPAdapter
from bs4 import BeautifulSoup

# 1. Configurar Session con Pool grande (elimina todos los WARNING de urllib3)
session = requests.Session()
adapter = HTTPAdapter(pool_connections=32, pool_maxsize=32, max_retries=2)
session.mount("https://", adapter)
session.mount("http://", adapter)
session.headers.update({
    "User-Agent": "Mozilla/5.0 (X11; Arch Linux; Linux x86_64; rv:135.0) Gecko/20100101 Firefox/135.0"
})

new_raw_documents = []

# =========================================================================
# 2. EXTRACCIÓN 100% COMPLETA VÍA GIT PARA HYPRLAND, NIRI Y SDKs DE MCP (Python/TS)
#    (Saltea Cloudflare, cero límite de páginas y preserva etiquetas de código exactas)
# =========================================================================
GIT_DOC_REPOS = {
    "hyprland_wiki": {
        "repo_url": "https://github.com/hyprwm/hyprland-wiki.git",
        "folder": "repo_hyprland_wiki",
        "glob_pattern": "pages/**/*.md",
        "base_web": "https://wiki.hypr.land/"
    },
    "niri_docs_exact": {
        "repo_url": "https://github.com/niri-wm/niri.git",
        "folder": "repo_niri",
        "glob_pattern": "docs/wiki/**/*.md",
        "base_web": "https://niri-wm.github.io/niri/"
    },
    "mcp_python_sdk": {
        "repo_url": "https://github.com/modelcontextprotocol/python-sdk.git",
        "folder": "repo_mcp_py",
        "glob_pattern": "**/*.md",
        "base_web": "https://py.sdk.modelcontextprotocol.io/"
    },
    "mcp_typescript_sdk": {
        "repo_url": "https://github.com/modelcontextprotocol/typescript-sdk.git",
        "folder": "repo_mcp_ts",
        "glob_pattern": "**/*.md",
        "base_web": "https://ts.sdk.modelcontextprotocol.io/"
    }
}

print("📦 Clonando repositorios oficiales de documentación (Hyprland, Niri, MCP SDKs)...")
for src_name, r_cfg in GIT_DOC_REPOS.items():
    folder = r_cfg["folder"]
    if not os.path.exists(folder):
        os.system(f"git clone --depth 1 -q {r_cfg['repo_url']} {folder}")
    md_files = glob.glob(os.path.join(folder, r_cfg["glob_pattern"]), recursive=True)
    count = 0
    for fpath in md_files:
        try:
            with open(fpath, "r", encoding="utf-8", errors="ignore") as f:
                content = f.read().strip()
            if len(content) > 200:
                rel_name = os.path.relpath(fpath, folder)
                new_raw_documents.append({
                    "source": "niri_docs" if src_name == "niri_docs_exact" else src_name,
                    "url": r_cfg["base_web"] + rel_name,
                    "title": rel_name,
                    "content": content
                })
                count += 1
        except Exception:
            pass
    print(f"   ✅ [{src_name}] {count} archivos Markdown originales extraídos al 100%.")

# =========================================================================
# 3. API OFICIAL DE ARCHWIKI (System Administration + Wayland + Pacman + Systemd)
# =========================================================================
print("🏛️ Extrayendo ArchWiki vía API oficial de MediaWiki (Sin bloqueos HTML)...")
ARCH_CATEGORIES = [
    "Category:System_administration",
    "Category:Package_management",
    "Category:Systemd",
    "Category:Wayland",
    "Category:Security",
    "Category:Networking"
]

arch_titles = set()
for cat in ARCH_CATEGORIES:
    try:
        r = session.get(
            "https://wiki.archlinux.org/api.php",
            params={"action": "query", "list": "categorymembers", "cmtitle": cat, "cmlimit": 250, "format": "json"},
            timeout=15
        ).json()
        for m in r.get("query", {}).get("categorymembers", []):
            t = m.get("title", "")
            # Ignorar subcategorías o traducciones a otros idiomas
            if not t.startswith("Category:") and "(" not in t:
                arch_titles.add(t)
    except Exception:
        pass

print(f"   🔍 {len(arch_titles)} artículos técnicos únicos encontrados en ArchWiki. Descargando contenido...")

def fetch_archwiki_article(title):
    try:
        r = session.get(
            "https://wiki.archlinux.org/api.php",
            params={"action": "parse", "page": title, "prop": "text", "format": "json"},
            timeout=12
        ).json()
        html_raw = r.get("parse", {}).get("text", {}).get("*", "")
        if not html_raw:
            return None
        soup = BeautifulSoup(html_raw, "html.parser")
        for junk in soup.select(".mw-editsection, .toc, #toc, .catlinks, style, script, .archwiki-template-meta-related-articles-start"):
            junk.decompose()
        for pre in soup.find_all("pre"):
            code_txt = pre.get_text().strip()
            if code_txt:
                pre.replace_with(f"\n```bash\n{code_txt}\n```\n")
        for h_lvl in range(1, 5):
            for h in soup.find_all(f"h{h_lvl}"):
                h.replace_with(f"\n\n{'#' * h_lvl} {h.get_text(' ', strip=True)}\n")
        clean_txt = soup.get_text().strip()
        if len(clean_txt) > 300:
            return {
                "source": "arch_wiki_sysadmin",
                "url": f"https://wiki.archlinux.org/title/{title.replace(' ', '_')}",
                "title": f"ArchWiki: {title}",
                "content": clean_txt
            }
    except Exception:
        return None
    return None

arch_extracted = 0
with ThreadPoolExecutor(max_workers=6) as ex:
    futures = [ex.submit(fetch_archwiki_article, t) for t in list(arch_titles)[:320]]
    for fut in as_completed(futures):
        res = fut.result()
        if res:
            new_raw_documents.append(res)
            arch_extracted += 1
print(f"   ✅ [arch_wiki_sysadmin] {arch_extracted} artículos completos de ArchWiki extraídos vía API.")

# =========================================================================
# 4. API JSON OFICIAL DE ARCH LINUX PACKAGES (Paquetes clave del ecosistema)
# =========================================================================
print("📦 Consultando API JSON oficial de paquetes de Arch Linux...")
ARCH_PKG_QUERIES = [
    "wayland", "hyprland", "niri", "xdg-desktop-portal", "pipewire", "wireplumber",
    "systemd", "pacman", "qt6", "python", "rust", "lua", "clang", "cmake", "typescript",
    "nftables", "openssh", "btrfs", "micro", "nano", "jq", "ripgrep", "fd", "alacritty", "foot"
]
pkg_docs_lines = ["# Catálogo Oficial de Paquetes de Arch Linux (Core / Extra)\n"]
seen_pkgs = set()

for q in ARCH_PKG_QUERIES:
    try:
        r = session.get(f"https://archlinux.org/packages/search/json/?q={q}", timeout=10).json()
        for pkg in r.get("results", [])[:35]:
            pname = pkg.get("pkgname")
            if pname and pname not in seen_pkgs:
                seen_pkgs.add(pname)
                repo = pkg.get("repo", "extra")
                desc = pkg.get("pkgdesc", "")
                ver = pkg.get("pkgver", "")
                deps = ", ".join(pkg.get("depends", [])[:10])
                pkg_docs_lines.append(
                    f"## Paquete `{pname}` ({repo} - v{ver})\n"
                    f"- **Descripción:** {desc}\n"
                    f"- **Instalación en Arch:** `sudo pacman -S {pname}`\n"
                    f"- **Dependencias principales:** {deps}\n"
                )
    except Exception:
        pass

if len(seen_pkgs) > 0:
    new_raw_documents.append({
        "source": "arch_packages",
        "url": "https://archlinux.org/packages/",
        "title": "Arch Linux Official Packages Reference",
        "content": "\n".join(pkg_docs_lines)
    })
print(f"   ✅ [arch_packages] {len(seen_pkgs)} paquetes oficiales de Arch Linux documentados vía API JSON.")

# =========================================================================
# 5. QUICKSHELL COMPLETO (SIN LÍMITE DE 250) + QML/Qt6, C++, LUA 5.4, TS Y BASH
# =========================================================================
EXTRA_WEB_SOURCES = {
    "quickshell_v032_full": {
        "seeds": [
            "https://quickshell.org/docs/v0.3.2/",
            "https://quickshell.org/docs/v0.3.2/guide/",
            "https://quickshell.org/docs/v0.3.2/types/"
        ],
        "allowed_prefixes": ["https://quickshell.org/docs/"],
        "max_pages": 600,  # Sin el techo de 250 para bajar absolutamente todos los tipos QML
        "default_lang": "qml"
    },
    "qml_qt6_cpp_docs": {
        "seeds": [
            "https://doc.qt.io/qt-6/qmlreference.html",
            "https://doc.qt.io/qt-6/qtqml-syntax-basics.html",
            "https://doc.qt.io/qt-6/qtqml-javascript-expressions.html",
            "https://doc.qt.io/qt-6/qtqml-cppintegration-overview.html",
            "https://doc.qt.io/qt-6/qtqml-cppintegration-exposecppattributes.html",
            "https://doc.qt.io/qt-6/qtquick-qmlmodule.html"
        ],
        "allowed_prefixes": [
            "https://doc.qt.io/qt-6/qtqml-",
            "https://doc.qt.io/qt-6/qml-",
            "https://doc.qt.io/qt-6/qtquick-"
        ],
        "max_pages": 160,
        "default_lang": "qml"
    },
    "lua_54_manual": {
        "seeds": ["https://www.lua.org/manual/5.4/manual.html"],
        "allowed_prefixes": ["https://www.lua.org/manual/5.4/"],
        "max_pages": 10,
        "default_lang": "lua"
    },
    "gnu_bash_manual": {
        "seeds": ["https://www.gnu.org/software/bash/manual/bash.html"],
        "allowed_prefixes": ["https://www.gnu.org/software/bash/manual/"],
        "max_pages": 10,
        "default_lang": "bash"
    },
    "typescript_docs": {
        "seeds": [
            "https://www.typescriptlang.org/docs/handbook/2/basic-types.html",
            "https://www.typescriptlang.org/docs/handbook/2/everyday-types.html",
            "https://www.typescriptlang.org/docs/handbook/2/narrowing.html",
            "https://www.typescriptlang.org/docs/handbook/2/functions.html",
            "https://www.typescriptlang.org/docs/handbook/2/objects.html",
            "https://www.typescriptlang.org/docs/handbook/2/modules.html"
        ],
        "allowed_prefixes": ["https://www.typescriptlang.org/docs/handbook/"],
        "max_pages": 80,
        "default_lang": "typescript"
    }
}

def smart_html_to_md(soup, fallback_lang="text"):
    for junk in soup.select("nav, header, footer, aside, script, style, noscript, svg, .sidebar, .navbar, .toc"):
        junk.decompose()
    main_node = soup.select_one("main") or soup.select_one("article") or soup.select_one("div.context") or soup.body
    if not main_node:
        return ""
    for pre in main_node.find_all("pre"):
        # Detectar el lenguaje real desde la clase CSS del bloque <code> si existe
        code_el = pre.find("code")
        detected_lang = fallback_lang
        if code_el and code_el.get("class"):
            for cls in code_el.get("class"):
                if cls.startswith(("language-", "lang-")):
                    detected_lang = cls.split("-", 1)[1]
                    break
        code_txt = pre.get_text().strip()
        if code_txt:
            pre.replace_with(f"\n```{detected_lang}\n{code_txt}\n```\n")
    for h_lvl in range(1, 5):
        for h in main_node.find_all(f"h{h_lvl}"):
            h_txt = h.get_text(" ", strip=True)
            if h_txt:
                h.replace_with(f"\n\n{'#' * h_lvl} {h_txt}\n")
    return main_node.get_text().strip()

for s_name, cfg in EXTRA_WEB_SOURCES.items():
    visited = set(urldefrag(u)[0] for u in cfg["seeds"])
    queue = deque(visited)
    pages_out = []
    print(f"🕷️ [{s_name}] Extrayendo documentación recursiva...")
    with ThreadPoolExecutor(max_workers=10) as ex:
        while queue and len(pages_out) < cfg["max_pages"]:
            batch = [queue.popleft() for _ in range(min(12, len(queue)))]
            futs = {ex.submit(session.get, u, timeout=12): u for u in batch}
            for fut in as_completed(futs):
                u = futs[fut]
                try:
                    resp = fut.result()
                    if resp.status_code != 200 or "text/html" not in resp.headers.get("Content-Type", ""):
                        continue
                    soup = BeautifulSoup(resp.text, "html.parser")
                    title = soup.title.get_text(strip=True) if soup.title else u
                    for a in soup.find_all("a", href=True):
                        full_u = urldefrag(urljoin(u, a["href"]))[0]
                        if any(full_u.startswith(p) for p in cfg["allowed_prefixes"]):
                            if full_u not in visited and not EXCLUDE_URL_REGEX.search(full_u):
                                visited.add(full_u)
                                queue.append(full_u)
                    md = smart_html_to_md(soup, cfg["default_lang"])
                    if len(md) > 250:
                        target_src = "quickshell_v032" if s_name == "quickshell_v032_full" else s_name
                        pages_out.append({"source": target_src, "url": u, "title": title, "content": md})
                except Exception:
                    pass
    new_raw_documents.extend(pages_out)
    print(f"   ✅ [{s_name}] {len(pages_out)} páginas extraídas.")

# =========================================================================
# 6. UNIFICAR, DEDUPLICAR Y AUDITAR EL CORPUS COMPLETO
# =========================================================================
new_chunks = split_into_training_chunks(new_raw_documents)

# Cargar los bloques previos y fusionar sin duplicados (reemplazando niri_docs por la versión Git con etiquetas exactas)
existing_chunks = []
if os.path.exists("dataset_documentacion_real.jsonl"):
    with open("dataset_documentacion_real.jsonl", "r", encoding="utf-8") as f:
        for line in f:
            item = json.loads(line)
            if item["source"] not in ("niri_docs", "quickshell_v032"):
                existing_chunks.append(item)

merged_chunks = existing_chunks + new_chunks
unique_chunks = []
seen_hashes = set()

for ch in merged_chunks:
    # Deduplicar por los primeros 300 caracteres del cuerpo técnico
    body_sig = re.sub(r'\s+', ' ', ch["text"])[:350]
    if body_sig not in seen_hashes:
        seen_hashes.add(body_sig)
        unique_chunks.append(ch)

FINAL_CORPUS_FILE = "dataset_corpus_completo_auditado.jsonl"
with open(FINAL_CORPUS_FILE, "w", encoding="utf-8") as f:
    for ch in unique_chunks:
        f.write(json.dumps(ch, ensure_ascii=False) + "\n")

# Auditoría detallada por fuente
counts_by_src = Counter(ch["source"] for ch in unique_chunks)
chars_by_src = Counter()
code_blocks_by_src = Counter()
for ch in unique_chunks:
    chars_by_src[ch["source"]] += len(ch["text"])
    code_blocks_by_src[ch["source"]] += ch["text"].count("```") // 2

total_chars = sum(chars_by_src.values())
print("\n" + "=" * 88)
print("📊 AUDITORÍA TÉCNICA DEL DATASET UNIFICADO (Sin duplicados)")
print("=" * 88)
print(f"{'Fuente / Lenguaje':<24} | {'Bloques':<9} | {'Bloques Código':<15} | {'Volumen (MB)':<13} | {'% del Corpus'}")
print("-" * 88)
for src, cnt in sorted(counts_by_src.items(), key=lambda x: x[1], reverse=True):
    mb_s = chars_by_src[src] / (1024**2)
    pct = (chars_by_src[src] / total_chars) * 100 if total_chars else 0
    print(f"{src:<24} | {cnt:<9} | {code_blocks_by_src[src]:<15} | {mb_s:<13.2f} | {pct:.1f}%")
print("-" * 88)
print(f"{'TOTAL UNIFICADO':<24} | {len(unique_chunks):<9} | {sum(code_blocks_by_src.values()):<15} | {total_chars/(1024**2):<13.2f} | 100.0%")
````

````python
import re
import json
from collections import Counter

INPUT_PATH = "/content/dataset_entrenamiento_balanceado.jsonl"
CLEAN_OUTPUT_PATH = "/content/dataset_limpio_verificado.jsonl"

with open(INPUT_PATH, "r", encoding="utf-8") as f:
    raw_records = [json.loads(line) for line in f]

# Patrones de lenguajes/distros ajenos a tu stack que queremos fuera del dataset
FOREIGN_STACK_RE = re.compile(
    r'\b(sudo\s+apt|apt-get|sudo\s+dnf|dnf\s+install|yum\s+install|brew\s+install|'
    r'dotnet\s+run|dotnet\s+add|mvn\s+clean|gradle\s+build|composer\s+require|cargo\s+windows)\b',
    re.IGNORECASE
)

def sanitize_and_repair_record(rec):
    txt = rec.get("text", "")
    src = rec.get("source", "")

    # 1. Eliminar bloques de código marcados como inválidos a propósito (must-fail en Niri, compile_fail en Rust)
    txt = re.sub(
        r'```[a-zA-Z0-9_,-]*(?:must-fail|compile_fail|panics|error)[a-zA-Z0-9_,-]*\n.*?```',
        '',
        txt,
        flags=re.DOTALL | re.IGNORECASE
    )

    # 2. Si el bloque tiene un ``` abierto al final que quedó truncado por longitud, recortar la cola rota
    if txt.count("```") % 2 != 0:
        last_fence = txt.rfind("```")
        txt = txt[:last_fence].strip()

    # 3. Limpiar shortcodes de Hugo/MDX ({{< ... >}}, {% ... %}) y tags HTML residuales
    txt = re.sub(r'\{\{[<%].*?[>%]\}\}', '', txt, flags=re.DOTALL)
    txt = re.sub(r'</?(?:div|span|table|tr|td|th|thead|tbody|script|style|iframe|img|br|hr)[^>]*>', '', txt, flags=re.IGNORECASE)

    # 4. Normalizar etiquetas de bloques de código (quitar theme={null}, no-run, ignore, etc.)
    #    y reparar falsos ```json que en realidad contienen comandos bash
    def fix_code_block(match):
        header = match.group(1).strip()
        body = match.group(2).strip()
        if not body:
            return ""

        # Quitar atributos tipo theme={null} o comas (,no-run, ,ignore, ,edition2021)
        lang = header.split()[0].split(",")[0].lower() if header else "text"
        lang_map = {"sh": "bash", "shell": "bash", "console": "bash", "zsh": "bash", "ts": "typescript", "js": "javascript", "py": "python", "rs": "rust", "c++": "cpp"}
        lang = lang_map.get(lang, lang)

        # Si dice ```json pero NO empieza con { ni [ (ej. "cd carpeta" o "npm run build"), pasarlo a ```bash
        if lang == "json" and not body.lstrip().startswith(("{", "[")):
            if any(cmd in body for cmd in ("npm ", "npx ", "uv ", "pip ", "cd ", "mkdir ", "curl ", "git ", "python", "node ")):
                lang = "bash"
            else:
                return ""  # Descartar falso bloque JSON irrelevante

        return f"```{lang}\n{body}\n```"

    txt = re.sub(r'```([^\n]*)\n(.*?)```', fix_code_block, txt, flags=re.DOTALL)

    # 5. Filtrar líneas sueltas de otras distros (apt/dnf/brew) o tecnologías ajenas (dotnet/maven)
    if FOREIGN_STACK_RE.search(txt):
        # Si es un bloque de Niri/Hyprland que muestra cómo instalar en Fedora/Ubuntu/Arch, dejamos solo la parte de Arch
        lines = txt.splitlines()
        filtered_lines = [l for l in lines if not FOREIGN_STACK_RE.search(l)]
        txt = "\n".join(filtered_lines)
        if FOREIGN_STACK_RE.search(txt):
            return None

    # 6. Limpiar exceso de URLs crudas o caracteres corruptos
    if "\ufffd" in txt or len(re.findall(r'https?://', txt)) > 8:
        return None

    # 7. Exigir que el bloque tenga al menos 1 bloque de código válido y cerrado (o comandos reales de Arch)
    fence_count = txt.count("```")
    if fence_count < 2 or (fence_count % 2 != 0):
        return None

    # Limpiar espacios múltiples
    txt = re.sub(r'\n{3,}', '\n\n', txt).strip()
    if len(txt) < 180:
        return None

    return {
        "source": src,
        "title": rec.get("title", ""),
        "url": rec.get("url", ""),
        "text": txt
    }

sanitized_records = []
seen_clean_sigs = set()

for r in raw_records:
    cleaned = sanitize_and_repair_record(r)
    if cleaned:
        sig = re.sub(r'\s+', ' ', cleaned["text"])[:350]
        if sig not in seen_clean_sigs:
            seen_clean_sigs.add(sig)
            sanitized_records.append(cleaned)

with open(CLEAN_OUTPUT_PATH, "w", encoding="utf-8") as f:
    for r in sanitized_records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

# =========================================================================
# VERIFICACIÓN FORENSE POST-LIMPIEZA (Debe dar 0 en todos los errores)
# =========================================================================
post_issues = Counter()
for r in sanitized_records:
    t = r["text"]
    if t.count("```") % 2 != 0:
        post_issues["1_codigo_truncado_sin_cerrar"] += 1
    if re.search(r'(\{\{[<%].*?[>%]\}\}|</?(?:div|span|table|tr|td|script|style)\b)', t, re.I):
        post_issues["2_basura_html_o_hugo_shortcodes"] += 1
    if FOREIGN_STACK_RE.search(t):
        post_issues["3_comandos_otras_distros_o_dotnet"] += 1
    if t.count("```") == 0:
        post_issues["4_puro_texto_sin_codigo"] += 1
    if re.search(r'```[a-z0-9_,-]*(?:must-fail|compile_fail|theme=)', t, re.I):
        post_issues["5_etiquetas_must_fail_o_theme_null"] += 1

src_counts = Counter(r["source"] for r in sanitized_records)
print("=" * 80)
print(f"✨ AUDITORÍA POST-PURIFICACIÓN: {len(sanitized_records)} BLOQUES DE CÓDIGO 100% VERIFICADOS")
print("=" * 80)
print(f"   - Errores de código truncado sin cerrar    : {post_issues['1_codigo_truncado_sin_cerrar']}")
print(f"   - Basura HTML / Shortcodes Hugo            : {post_issues['2_basura_html_o_hugo_shortcodes']}")
print(f"   - Comandos apt/dnf/brew/dotnet infiltrados : {post_issues['3_comandos_otras_distros_o_dotnet']}")
print(f"   - Bloques sin código                       : {post_issues['4_puro_texto_sin_codigo']}")
print(f"   - Etiquetas must-fail / theme={{null}}       : {post_issues['5_etiquetas_must_fail_o_theme_null']}")
print("-" * 80)
print("📊 DISTRIBUCIÓN DE BLOQUES 100% PUROS CON CÓDIGO POR TECNOLOGÍA:")
for s, c in sorted(src_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"   • {s:<24}: {c:>4} bloques puros con código")
print("=" * 80)
print(f"💾 Guardado en: {CLEAN_OUTPUT_PATH}")
````
